```python
df['join_title'] = df['join_title'].str.lower().str.strip()
filtered_df_books['Name'] = filtered_df_books['Name'].str.lower().str.strip()
```

```python
# Clean the titles to make sure they match
df['join_title'] = df['join_title'].str.lower().str.strip()
filtered_df_books['Name'] = filtered_df_books['Name'].str.lower().str.strip()

# Keep only the rows in df where join_title is in the books database
df_books_only = df[df['join_title'].isin(filtered_df_books['Name'])].copy()
```

```python
merged_df = df.merge(
    filtered_df_books[['Name', 'Publisher', 'Pages', 'Rating', 'RatingDistTotal', 'Authors']],
    left_on='join_title',
    right_on='Name',
    how='left',
    suffixes=('', '_book')
)
print(merged_df.columns)
column_mapping = {
    'Publisher': 'book_publisher',
    'Pages': 'book_pages',
    'Rating': 'book_rating',
    'RatingDistTotal': 'book_ratings_count',
    'Authors': 'join_author'
}
for book_col, df_col in column_mapping.items():
    book_col_name = f'{book_col}_book'
    if book_col_name in merged_df.columns:
        merged_df[df_col] = merged_df[df_col].combine_first(merged_df[book_col_name])
merged_df = merged_df.drop(columns=[c for c in merged_df.columns if c.endswith('_book')] + ["Name"])
merged_df
```

```python
merged_df = df_books_only.merge(
    filtered_df_books[['Name', 'Publisher', 'Pages', 'Rating', 'RatingDistTotal', 'Authors']],
    left_on='join_title',
    right_on='Name',
    how='left',
    suffixes=('', '_book')
)

column_mapping = {
    'Publisher': 'book_publisher',
    'Pages': 'book_pages',
    'Rating': 'book_rating',
    'RatingDistTotal': 'book_ratings_count',
    'Authors': 'join_author'
}

for book_col, df_col in column_mapping.items():
    book_col_name = f'{book_col}_book'
    if book_col_name in merged_df.columns:
        merged_df[df_col] = merged_df[df_col].combine_first(merged_df[book_col_name])

merged_df = merged_df.drop(columns=[c for c in merged_df.columns if c.endswith('_book')] + ['Name'])
df = merged_df  # now df contains only books with filled info
df
```
