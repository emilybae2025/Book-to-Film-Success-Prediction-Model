```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("final_data.csv")
df
```

```python
# Remove the unnamed column if it exists
if 'Unnamed: 0' in df.columns:
    df = df.drop('Unnamed: 0', axis=1)
df.head()
```

```python
df = df.dropna(subset=['Book Rating', 'Movie Rating'])
```

```python
model = LinearRegression()
ind = ["Book Rating"]
dep = "Movie Rating"
model.fit(df[ind], df[dep])
df["Movie Rating Prediction"] = model.predict(df[ind])
df
```

```python
df["Movie Rating Residual"] = df["Movie Rating"] - df["Movie Rating Prediction"]
df.head()
```

<p>Linear Regression predicts Movie Rating very well based on Book Rating.</p>

```python
df = pd.read_csv("final_data.csv")
# Remove the unnamed column if it exists
if 'Unnamed: 0' in df.columns:
    df = df.drop('Unnamed: 0', axis=1)
df = df.fillna(0)
df.head()
```

```python
model = LinearRegression()

ind = ["Book Rating"]
dep = "movie_revenue"
model.fit(df[ind], df[dep])

df["Movie Rev Prediction"] = model.predict(df[ind])
df["Movie Rev Residual"] = df["movie_revenue"] - df["Movie Rev Prediction"]
df.head()
```

<p>Linear Regression doesn't do a good job at predicting Movie Revenue based on book rating. This is clear in the residual since it is super high</p>
