```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("cleaned_df.csv")
df
```

```python
# Remove the unnamed column if it exists
if 'Unnamed: 0' in df.columns:
    df = df.drop('Unnamed: 0', axis=1)
df
```

```python
sns.histplot(df['Rating'])
```

```python

plt.figure(figsize=(10,6))
plt.scatter(df['Rating'], df['movie_revenue']/1e6)
plt.yscale('log')
plt.xlabel("Book Rating")
plt.ylabel("Movie Revenue (Millions, log scale)")
plt.title("Book Rating vs Movie Revenue")
plt.show()
# it appears that the  movie reveue is better when book rating is better.
```

```python
# book rating vs movie rating (scatterplot)

# create new columns for percent of rating (since movie is out of 10 and book is out of 5)
df["movie_rating_percent"] = df["imdb_rating"] / 10
df["book_rating_percent"] = df["Rating"] / 5
# remove ratings of 0 (no rating)
df = df[(df.movie_rating_percent != 0) & (df.book_rating_percent != 0)]
# scatter plot w/o regression line
df.plot.scatter(title="Book Ratings vs. Movie Ratings Percentages", x="movie_rating_percent", y="book_rating_percent", xlabel="Movie Rating", ylabel="Book Rating")
# scatter plot w regression line
sns.regplot(data=df, x="movie_rating_percent", y="book_rating_percent", ci=None, line_kws={'color':'red'})
plt.title("Book Ratings Percentages vs. Movie Ratings Percentages")
```

```python
# movie budget vs revenue
df = df[(df.movie_revenue != 0) | (df.movie_budget != 0)] # remove revenue and budget values of 0

sns.regplot(data=df, x="movie_budget", y="movie_revenue", ci=None, line_kws={'color':'red'})
plt.title("Movie Budget vs. Movie Revenue")

# zoom in to cluster
plt.xlim(0, 0.4e9) 
plt.ylim(-0.5e9, 3e9) 

plt.show()
```

```python
# movie profit vs. rating
# new column profit = revenue - budget
df["movie_profit"] = df.movie_revenue - df.movie_budget

# scatterplot
sns.regplot(data=df, x="movie_profit", y="movie_rating_percent", ci=None, line_kws={'color':'red'})
plt.title("Movie Profit vs. Movie Ratings Percentages")

# zoom in to cluster
plt.xlim(-1e8, 2e8) 
plt.ylim(0, 1) 

plt.show()
```
