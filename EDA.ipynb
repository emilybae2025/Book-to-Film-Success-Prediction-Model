```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
original = pd.read_csv("cleaned_df.csv")
```

```python
df = pd.read_csv("cleaned_df.csv")
```

```python
# Remove the unnamed column if it exists
if 'Unnamed: 0' in df.columns:
    df = df.drop('Unnamed: 0', axis=1)
df
```

```python
df["Book Rating"] = df["Rating"]/5
df["Movie Rating"] = df["imdb_rating"]/10
df.head()
```

```python
df["movie_release"] = df["movie_release"].astype(int)
df["RatingDistTotal"] = df["RatingDistTotal"].str[6:]
df["imdb_id"] = df["imdb_id"].str[2:]
df.head()
```

```python
df['movie_budget'] = df['movie_budget'].replace(0, np.nan)
df['Pages'] = df['Pages'].replace(0, np.nan)
df[(df.Pages == 0) | (df.movie_budget == 0)]
```

```python
df['movie_release'] = df['movie_release'].replace(0, np.nan)
df['Book Rating'] = df['Book Rating'].replace(0, np.nan)
df[(df.movie_release == 0) | (df['Book Rating'] == 0)]
```

```python
df["imdb_rating"] = df["imdb_rating"].replace(0, np.nan)
df[df.imdb_rating == 0]
```

```python
df = df.drop(["Rating", "imdb_id", "join_title", "imdb_rating"], axis=1)
df
```

```python
sns.histplot(df['Book Rating'])
```

```python

plt.figure(figsize=(10,6))
plt.scatter(df['Book Rating'], df['movie_revenue']/1e6)
plt.yscale('log')
plt.xlabel("Book Rating")
plt.ylabel("Movie Revenue (Millions, log scale)")
plt.title("Book Rating vs Movie Revenue")
plt.show()
# it appears that the  movie reveue is better when book rating is better.
```

```python
# book rating vs movie rating (scatterplot)

# scatter plot w/o regression line
df.plot.scatter(title="Book Ratings vs. Movie Ratings Percentages", x="Movie Rating", y="Book Rating", xlabel="Movie Rating", ylabel="Book Rating")
# scatter plot w regression line
sns.regplot(data=df, x="Movie Rating", y="Book Rating", ci=None, line_kws={'color':'red'})
plt.title("Book Ratings Percentages vs. Movie Ratings Percentages")
```

```python
df.plot.scatter(title="Book Ratings vs. Movie Ratings Percentages", y="Movie Rating", x="Book Rating", xlabel="Movie Rating", ylabel="Book Rating")
sns.regplot(data=df, x="Movie Rating", y="Book Rating", ci=None, line_kws={'color':'red'})
plt.title("Book Ratings Percentages vs. Movie Ratings Percentages")
```

```python
# movie budget vs revenue
sns.regplot(data=df, x="movie_budget", y="movie_revenue", ci=None, line_kws={'color':'red'})
plt.title("Movie Budget vs. Movie Revenue")

# zoom in to cluster
plt.xlim(0, 0.4e9) 
plt.ylim(-0.5e9, 3e9) 

plt.show()
```

```python
# movie profit vs. rating
# new column profit = revenue - budget
df["movie_profit"] = df.movie_revenue - df.movie_budget

# scatterplot
sns.regplot(data=df, x="movie_profit", y="Movie Rating", ci=None, line_kws={'color':'red'})
plt.title("Movie Profit vs. Movie Ratings Percentages")

# zoom in to cluster
plt.xlim(-1e8, 2e8) 
plt.ylim(0, 1) 

plt.show()
```
